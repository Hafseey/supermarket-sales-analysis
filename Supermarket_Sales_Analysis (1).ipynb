```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

kpi_values = [
    df['Sales'].sum(),
    df['Sales'].mean(),
    df['Sales'].max(),
    df['Quantity'].sum()
]

kpi_titles = [
    'Total Sales',
    'Average Sale',
    'Highest Sale',
    'Total Quantity Sold'
]

for ax, value, title in zip(axes, kpi_values, kpi_titles):
    ax.text(
        0.5, 0.5,
        f'{value:,.2f}',
        ha='center',
        va='center',
        fontsize=20,
        fontweight='bold'
    )

    ax.set_title(title)
    ax.axis('off')

plt.suptitle('Supermarket Sales — Key Performance Indicators', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# 1. Sales by Product Line
product_sales = df.groupby('Product line')['Sales'].sum().sort_values(ascending=False)
axes[0, 0].bar(product_sales.index, product_sales.values)
axes[0, 0].set_title('Sales by Product Line')
axes[0, 0].tick_params(axis='x', rotation=45)

# 2. Sales by Branch
branch_sales = df.groupby('Branch')['Sales'].sum()
axes[0, 1].bar(branch_sales.index, branch_sales.values)
axes[0, 1].set_title('Sales by Branch')

# 3. Monthly Sales Trend
monthly_sales = df.groupby('Month')['Sales'].sum()
axes[1, 0].plot(monthly_sales.index, monthly_sales.values, marker='o')
axes[1, 0].set_title('Monthly Sales Trend')
axes[1, 0].set_xlabel('Month')
axes[1, 0].set_ylabel('Sales')

# 4. Sales by Payment Method
payment_sales = df.groupby('Payment')['Sales'].sum()
axes[1, 1].bar(payment_sales.index, payment_sales.values)
axes[1, 1].set_title('Sales by Payment Method')

plt.suptitle(
    'Supermarket Sales Performance Dashboard',
    fontsize=18,
    fontweight='bold'
)

plt.tight_layout()
plt.show()
```

```python
from matplotlib.ticker import FuncFormatter

fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# Format large numbers with commas
formatter = FuncFormatter(lambda x, pos: f'{x:,.0f}')

# 1. Sales by Product Line
product_sales = df.groupby('Product line')['Sales'].sum().sort_values(ascending=False)
axes[0, 0].bar(product_sales.index, product_sales.values)
axes[0, 0].set_title('Sales by Product Line')
axes[0, 0].tick_params(axis='x', rotation=45)
axes[0, 0].yaxis.set_major_formatter(formatter)

# 2. Sales by Branch
branch_sales = df.groupby('Branch')['Sales'].sum()
axes[0, 1].bar(branch_sales.index, branch_sales.values)
axes[0, 1].set_title('Sales by Branch')
axes[0, 1].yaxis.set_major_formatter(formatter)

# 3. Monthly Sales Trend
monthly_sales = df.groupby('Month')['Sales'].sum()
axes[1, 0].plot(monthly_sales.index, monthly_sales.values, marker='o')
axes[1, 0].set_title('Monthly Sales Trend')
axes[1, 0].set_xlabel('Month')
axes[1, 0].set_ylabel('Sales')
axes[1, 0].yaxis.set_major_formatter(formatter)

# 4. Sales by Payment Method
payment_sales = df.groupby('Payment')['Sales'].sum()
axes[1, 1].bar(payment_sales.index, payment_sales.values)
axes[1, 1].set_title('Sales by Payment Method')
axes[1, 1].yaxis.set_major_formatter(formatter)

plt.suptitle(
    'Supermarket Sales Performance Dashboard',
    fontsize=18,
    fontweight='bold'
)

plt.tight_layout()
plt.show()
```
